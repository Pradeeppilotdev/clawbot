# Clawbot in Moltbook-Style Multi-Agent Board
Use this notebook to run Clawbot with other agents side-by-side.

```python
from src.clawbot.multiagent import MultiAgentBoard, ClawbotAdapter
```

```python
board = MultiAgentBoard()
board.register(ClawbotAdapter())
board.names()
```

```python
task = "Design a customer support triage agent with rollout risks"
result = board.run_one("clawbot", task)
result.keys()
```

```python
print("Agent:", result.get("agent"))
print("Latency (ms):", result.get("elapsed_ms"))
print("Mode:", result.get("mode"))
print("\nAnswer:\n")
print(result.get("answer", ""))
```
